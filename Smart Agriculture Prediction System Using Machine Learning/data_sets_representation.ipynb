```python
import pandas as pd

df = pd.read_csv("Crop_recommendation.csv")

print("RAW DATA")
display(df.head(10))

print("Shape:", df.shape)
```

```python
import warnings
warnings.filterwarnings("ignore")

import pickle
import numpy as np
import pandas as pd

# Load the same scalers used by app.py
with open("minmaxscaler.pkl", "rb") as f:
    mx = pickle.load(f)

with open("standscaler.pkl", "rb") as f:
    sc = pickle.load(f)

features = [
    "N", "P", "K",
    "temperature", "humidity",
    "ph", "rainfall"
]

# Convert to NumPy array, just like app.py
X = df[features].values

# EXACT SAME ORDER AS app.py
X_minmax = mx.transform(X)
X_preprocessed = sc.transform(X_minmax)

preprocessed_df = pd.DataFrame(
    X_preprocessed,
    columns=features
)

print("PREPROCESSED DATA")
display(preprocessed_df.head(10))
```
